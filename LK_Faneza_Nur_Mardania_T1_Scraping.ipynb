# Scraping Aplikasi Alfagift

```python
!pip install google-play-scraper

from google_play_scraper import reviews, Sort
import pandas as pd

# Aplikasi Alfagift
result, _ = reviews(
    'com.alfamart.alfagift',          # app ID dari URL Play Store
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=5000
)

data = [{
    'id': r['reviewId'],
    'teks': r['content'],
    'tanggal': r['at'],
    'sumber': 'playstore'
} for r in result]

df = pd.DataFrame(data)
df.to_csv('scraped_playstore2.csv', index=False)
print('Total review:', len(df))
```
