```python
import pandas as pd
import re

df = pd.read_csv('/content/drive/MyDrive/Data Analis Angkatan 2 Tahun 2026_Faneza Nur Mardania/21 Agustus 2026_Pertemuan 26/File Tugas/data_labeled.csv')   # dari Tugas 2

print('Dimensi data:', df.shape)
print('\n5 baris pertama:')
print(df.head())
print('\nDistribusi sentimen:')
print(df['sentimen'].value_counts())

```

```python
def clean_text(text):
    if not isinstance(text, str):
        return ''
    text = text.lower()                                  # case folding
    text = re.sub(r'https?://\S+|www\.\S+', '', text)  # hapus URL
    text = re.sub(r'@\w+|#\w+', '', text)               # hapus mention/hashtag
    text = re.sub(r'\d+', '', text)                      # hapus angka
    text = re.sub(r'[^\w\s]', '', text)                 # hapus tanda baca
    text = re.sub(r'\s+', ' ', text).strip()             # spasi berlebih
    return text

df['teks_clean'] = df['teks'].apply(clean_text)
print(df[['teks', 'teks_clean']].head())
```

```python
import pandas as pd
import re

!pip install Sastrawi

from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# ==========================================
# 1. INISIALISASI STEMMER DAN STOPWORD
# ==========================================

stemmer = StemmerFactory().create_stemmer()
factory = StopWordRemoverFactory()

# Stopword bawaan Sastrawi
stopwords = set(factory.get_stop_words())

# Kata negasi yang harus dipertahankan
negasi = {'tidak', 'bukan', 'belum', 'jangan', 'kurang', 'tanpa'}

# Hapus kata negasi dari daftar stopword
stopwords = stopwords - negasi

# Tambahan stopword
custom_stopwords = {
    'nya', 'nih', 'sih', 'lah', 'dong', 'kok', 'pun',
    'di', 'ke', 'dari', 'dan', 'yang', 'untuk',
    'dengan', 'pada', 'dalam', 'itu', 'ini', 'atau',
    'saya', 'aku', 'kami', 'kamu',
    'aja', 'saja', 'deh', 'yah', 'yaa','woi'
}

# Gabungkan stopword
all_stopwords = stopwords.union(custom_stopwords)


kamus_normalisasi = {
    'ga': 'tidak',
    'gak': 'tidak',
    'gk': 'tidak',
    'nggak': 'tidak',
    'engga': 'tidak',
    'enggak': 'tidak',
    'tak': 'tidak',
    'gabisa': 'tidak bisa',
    'gakbisa': 'tidak bisa',
    'gada': 'tidak ada',
    'gaada': 'tidak ada',
    'gausah': 'tidak usah',
    'gajelas': 'tidak jelas',
    'jgn': 'jangan',
    'yg': 'yang',
    'bgt': 'banget',
    'bangett': 'banget',
    'bangeet': 'banget',
    'smpe': 'sampai',
    'smp': 'sampai',
    'kalo': 'kalau',
    'kl': 'kalau',
    'udh': 'sudah',
    'udah': 'sudah',
    'blm': 'belum',
    'tdk': 'tidak',
    'tdak': 'tidak',
    'dlu': 'dulu',
    'skrg': 'sekarang',
    'tp': 'tapi',
    'tpi': 'tapi',
    'cm': 'cuma',
    'cuman': 'cuma',
    'jg': 'juga',
    'karna': 'karena',
    'krn': 'karena',
    'klo': 'kalau',
    'kalo': 'kalau',
    'mna': 'mana',
    'membatu' : 'membantu',
    'bnyak': 'banyak',
    'bgt2': 'banget',
    'ampe': 'sampai',
    'sampe': 'sampai',
    'pake': 'pakai',
    'dipake' : 'dipakai',
    'pakaii': 'pakai',
    'pesen': 'pesan',
    'pesenan': 'pesanan',
    'pengirimn': 'pengiriman',
    'peoduk': 'produk',
    'produkny': 'produk',
    'syaa': 'saya',
    'sy': 'saya',
    'gue': 'saya',
    'gueeeee' : 'saya',
    'gua': 'saya',
    'gw': 'saya',
    'gwe': 'saya',
    'biasae': 'biasanya',
    'slalu': 'selalu',
    'allhamdulillah': 'alhamdulillah',
    'mreka': 'mereka',
    'sbgai': 'sebagai',
    'sbg': 'sebagai',
    'kmu': 'kamu',
    'org': 'orang',
    'dr': 'dari',
    'dgn': 'dengan',
    'utk': 'untuk',
    'krg': 'kurang',
    'pd': 'pada',
    'pdhl': 'padahal',
    'pdh': 'padahal',
    'krna': 'karena',
    'spt': 'seperti',
    'blanja': 'belanja',
    'blm': 'belum',
    'bln': 'bulan',
    'br': 'baru',
    'hrs': 'harus',
    'hrus': 'harus',
    'bs': 'bisa',
    'bsa': 'bisa',
    'bisaa': 'bisa',
    'bgt': 'banget',
    'bgtt': 'banget',
    'lg': 'lagi',
    'lgi': 'lagi',
    'aja': 'saja',
    'doang': 'saja',
    'kali2': 'kali kali',
    'wkwkw': '',
    'wkwk': '',
    'wk': '',
    'yaa': '',
    'yah': '',
    'nih': '',
    'sih': '',
    'dong': '',
    'deh': '',
    'kok': '',
    'lah': '',
    'pun': '',
    'ny': 'nya',
    'apk': 'aplikasi',
    'apl': 'aplikasi',
    'app': 'aplikasi',
    'hp': 'handphone',
    'telp': 'telepon',
    'notelp': 'nomor telepon',
    'wa': 'whatsapp',
    'cs': 'customer service',
    'sop': 'prosedur',
    'respon': 'respons',
    'scra': 'secara',
    'ak': 'akan',
    'akn': 'akan',
    'msk': 'masuk',
    'maks': 'maksimal',
    'kmrn': 'kemarin',
    'tgl': 'tanggal',
    'jwbnnya': 'jawabannya',
    'ttp': 'tetap',
    'sdh': 'sudah'
}

def preprocess_full(text):
    text = clean_text(text)

    # Normalisasi
    words = text.split()
    words = [kamus_normalisasi.get(word, word) for word in words]

    # Stopword removal
    words = [word for word in words if word not in all_stopwords]
    text = ' '.join(words)

    # Stemming
    text = stemmer.stem(text)

    # Hapus stopword lagi setelah steamming
    words = text.split()
    words = [word for word in words if word not in all_stopwords]
    text = ' '.join(words)
    return text

df['teks_processed'] = df['teks_clean'].apply(preprocess_full)

print(df[['teks_clean', 'teks_processed']].head())

df.to_csv('data_preprocessed.csv', index=False)
```
